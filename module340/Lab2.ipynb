```python
import numpy as np
import matplotlib.pyplot as plt
```

## Part A — One-qubit Hamiltonian

```python
Z = np.array([[1, 0],
              [0,-1]], dtype=complex)

eigenvalues, eigenvectors = np.linalg.eigh(Z)

print("Eigenvalues:", eigenvalues)
print("Ground-state energy:", eigenvalues[0])
print("Ground-state vector:", eigenvectors[:, 0])
```

**Expected:** eigenvalues \(-1\) and \(+1\). The lowest is the ground-state energy.

## Part B — A state that is not an eigenstate

```python
ket_plus = np.array([1, 1], dtype=complex) / np.sqrt(2)

transformed = Z @ ket_plus
print("Z|+> =", transformed)

# Test whether the output is a scalar multiple of the input.
ratios = transformed / ket_plus
print("component-wise ratios:", ratios)
```

The two ratios differ, so \(|+\rangle\) is not an eigenstate of \(Z\). It is a valid quantum state, but it does not have one definite \(Z\)-energy.

## Part C — Expectation value

```python
expectation = np.vdot(ket_plus, Z @ ket_plus).real
print("<+|Z|+> =", expectation)
```
